# 00 — Exploratory Data Analysis (EDA)
### Diabetic Retinopathy Severity Grading (APTOS 2019 & Messidor-2)

This notebook analyzes:
1. Severity class distributions across APTOS 2019 and Messidor-2
2. Severe class imbalance characterization (No DR ~50%, Severe DR ~5%)
3. Image resolution variance, aspect ratios, and visual inspection per grade

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
GRADE_LABELS = {0: 'No DR', 1: 'Mild', 2: 'Moderate', 3: 'Severe', 4: 'Proliferative DR'}
print("EDA environment configured.")
```

### 1. Load Dataset Metadata

```python
aptos_csv = "../data/aptos2019/train.csv"
if os.path.exists(aptos_csv):
    df_aptos = pd.read_csv(aptos_csv)
    df_aptos['grade_name'] = df_aptos['diagnosis'].map(GRADE_LABELS)
    print(f"Loaded {len(df_aptos)} APTOS samples.")
    display(df_aptos.head())
else:
    print("APTOS data not downloaded locally yet. (Proceed with Colab cloud data)")
```

### 2. Plot Severity Grade Imbalance Distribution

```python
if 'df_aptos' in locals():
    plt.figure(figsize=(8, 4.5))
    palette = ['#10B981', '#3B82F6', '#F59E0B', '#EF4444', '#7C3AED']
    ax = sns.countplot(data=df_aptos, x='grade_name', palette=palette, order=list(GRADE_LABELS.values()))
    plt.title('APTOS 2019 Diagnosis Distribution', fontsize=14, fontweight='bold')
    plt.xlabel('Diabetic Retinopathy Grade')
    plt.ylabel('Patient Count')
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}', (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='baseline', fontsize=11, color='black', xytext=(0, 5),
                    textcoords='offset points')
    plt.tight_layout()
    plt.show()
```
