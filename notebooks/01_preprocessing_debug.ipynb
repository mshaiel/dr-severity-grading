# 01 — Ben Graham Preprocessing Pipeline Debug & Validation

This notebook verifies the multi-stage Ben Graham preprocessing:
1. Circular ROI bounding detection & border cropping
2. Aspect ratio preserving resize to 380x380
3. Contrast Limited Adaptive Histogram Equalization (CLAHE)
4. Local mean subtraction via Gaussian blurring and cv2.addWeighted
5. Side-by-side visualization for portfolio showcase

```python
import sys
sys.path.append('..')

import cv2
import matplotlib.pyplot as plt
import numpy as np
from src.data.preprocessing import ben_graham_pipeline
print("Preprocessing module imported successfully.")
```

### Step 1: Run 3-Stage Preprocessing Pipeline on Synthetic / Sample Fundus Image

```python
# Create synthetic test image
dummy = np.zeros((400, 400, 3), dtype=np.uint8)
cv2.circle(dummy, (200, 200), 170, (40, 80, 210), -1)
cv2.circle(dummy, (140, 200), 25, (100, 220, 240), -1)

intermediates = ben_graham_pipeline(dummy, image_size=380, return_intermediates=True)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(intermediates['raw'])
axes[0].set_title("① Raw Input Image")
axes[0].axis('off')

axes[1].imshow(intermediates['cropped'])
axes[1].set_title("② Circle Cropped & Resized (380x380)")
axes[1].axis('off')

axes[2].imshow(intermediates['clahe_normalized'])
axes[2].set_title("③ Ben Graham CLAHE + Local Mean (Model Input)")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
